<hr>

rubric={points}

The following questions pertain to the lecture on survival analysis. We'll consider the use case of customer churn analysis.

1. What is the problem with simply labeling customers are "churned" or "not churned" and using standard supervised learning techniques?
2. Consider customer A who just joined last week vs. customer B who has been with the service for a year. Who do you expect will leave the service first: probably customer A, probably customer B, or we don't have enough information to answer? Briefly explain your answer. 
3. If a customer's survival function is almost flat during a certain period, how do we interpret that?

<div class="alert alert-warning">

Solution_1
    
</div>

_Points:_ 6

1. Simply labeling customers as “churned” or “not churned” ignores time-to-event information and censoring. Many customers have not churned yet, but we don’t know when (or if) they will. Standard supervised learning drops these customers or incorrectly assumes they won’t churn.
2. We don’t have enough information to answer. In survival analysis, past survival changes risk (hazard), but without the survival/hazard function, we cannot definitively say who will churn first. It depends on how churn risk evolves over time.
3. If a customer’s survival function is flat over a period, it means the probability of churn during that time is very low (or near zero), very unlikely to leave during that interval.

## Can We Predict Credit Card Default? A Simple Analysis to Understand Risk
### Contents of this Blog

In this post, I walk through how I approached predicting credit card default using a structured machine learning pipeline. The blog covers:

Problem definition and data understanding
Exploratory data analysis
Feature engineering
Model creation
Results and interpretation
Limitations and possible improvements

### Problem Definition and Data Understanding

The goal of this project is to predict whether a customer will default on their credit card payment next month. This is a practical and important problem, as financial institutions rely on these predictions to manage risk and make lending decisions.

The dataset includes information about

1. Credit limit (LIMIT_BAL)
2. Repayment history over multiple months
3. Bill amounts and payment amounts
4. Demographic features

One important observation early on is that the dataset is imbalanced — only about 22% of customers default. This means a model could appear accurate while completely failing to identify high-risk customers, so evaluation metrics need to reflect this imbalanc

```python
from IPython.display import Image

Image("img/train.png")
```

### Exploratory Data Analysis

I begin by exploring the distribution of the target variable and key numerical features.

As shown in Figure 1, most customers do not default. This imbalance immediately suggests that accuracy is not a reliable metric, since predicting “no default” for all customers would still perform well on paper.

Looking at the distribution of credit limits (Figure 2), the data is heavily right-skewed. Most customers have moderate credit limits, while a small number have very large limits. This indicates the presence of outliers and supports the decision to scale numerical features.

These initial observations help guide me through both preprocessing decisions and model evaluation strategy.

### Feature Engineering

To better represent customer behavior, I created additional features based on existing variables.

In particular, I computed Average bill amount across months and Average payment amount across months.

These features capture overall financial patterns rather than treating each month independently. This is useful because default behavior is likely influenced by consistent trends rather than isolated events.

I also grouped features into numerical and categorical types to ensure they were processed appropriately within the modeling pipeline.

### Model Creation

I built a pipeline that combines preprocessing and modeling into a single workflow. This ensures consistent transformations across training and validation.

To establish a baseline, I first used a DummyClassifier. As expected, it performed poorly, but it provided a useful reference point.

I then trained several models:

Logistic regression
Decision tree
Random forest
Gradient boosting

Each model was evaluated using cross-validation and compared using F1-score, which balances precision and recall in the presence of class imbalance.



```python
from IPython.display import Image

Image("img/credit-limit.png")
```

### Results and Interpretation

In the end, the logistic regression model achieved an F1-score of approximately 0.48, which represents a clear improvement.

Diving deeper into the results in Figure 3, I discovered that more complex models such as random forests and gradient boosting performed slightly better, but the improvement was minimal. This suggests that the relationships in the data are relatively simple and largely linear. 

However, the decision tree performed very well on training data but much worse on validation data which could indicate overfitting, where the model captures noise rather than generalizable patterns. This meaning that my model wouldn't generalize well.

feature engineering also had a noticeable impact on performance. Aggregating bill and payment information into averages helped capture overall financial behavior more effectively. Showcasing again the importance of data preprocessing in data analysis.

Overall, the results suggest that increasing model complexity does not significantly improve performance, and that better features may be more valuable than more advanced models.

### Most importantly, the project highlights that:

Evaluation metrics matter, especially with imbalanced data
Simpler models can perform surprisingly well
Feature quality is often more important than model complexity

### Limitations and Possible Improvements


1. the dataset is limited to financial and repayment variables. It does not include behavioral or external factors that may influence default risk, such as spending habits or economic conditions.

2. the class imbalance makes it difficult for the model to accurately identify default cases. Even with F1-score, the model may still favor the majority class.

3. some models showed clear signs of overfitting. For example, the decision tree performed well on training data but did not generalize well, suggesting it captured noise rather than meaningful patterns.

4. the data represents a static snapshot in time. Customer behavior may change, and the model may not generalize well to future data without retraining.

One potential improvement would be to incorporate additional features, such as transaction-level data or behavioral metrics. These could provide deeper insight into customer risk and improve predictive performance.
